```python
#fill missing values in delivery_day using median
df['Delivery_Days'] = (
    df['Delivery_Days']
    .astype(str)
    .str.replace('days', '', regex=False)
    .str.strip()
)

df['Delivery_Days'] = pd.to_numeric(df['Delivery_Days'], errors='coerce')

df['Delivery_Days'].fillna(df['Delivery_Days'].median(), inplace=True)
```

```python
df['Delivery_Days']
```

```python
#replace misising coupon_code with no Coupon
df['Coupon_Code'].fillna("No Coupon")
```

```python
#part 4 string cleaning
#remove leading and trailing spaces in all string columns
str_cols = df.select_dtypes(include=['object'])

df[str_cols.columns] = str_cols.apply(lambda x: x.str.strip())
```

```python
df.columns
```

```python
#convert payment status to lowercase
df['Payment_Status']=df['Payment_Status'].str.lower()
```

```python
df['Payment_Status']
```

```python
#standardize values in Customer_city
df['Customer_City'] = df['Customer_City'].str.strip().str.lower().str.title()
```

```python
df['Customer_City']
```

```python
#clean coupon code(remove extra spaces and make uppercase)
df["Coupon_Code"]=df["Coupon_Code"].str.strip().str.upper()
```

```python
#numpy import
#replace unknown values with NaN
import numpy as np
df.replace("unknown", np.nan, inplace=True)
```

```python
#replace None and   NONE with NaN

df.replace(['NONE', 'none'], np.nan, inplace=True)
```

```python
#convert delivery days to numeric remove days
df['Delivery_Days'] = (
    df['Delivery_Days']
    .astype(str)
    .str.replace('days', '', regex=False)
    .str.strip()
)

df['Delivery_Days'] = pd.to_numeric(df['Delivery_Days'], errors='coerce')
```

```python
#convert discount percent into numeric(reemove%)
#(regex= normal string, coerce=its not show errors replaced by nan)
df['Discount_Percent'] = (
    df['Discount_Percent']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.strip()
)

df['Discount_Percent'] = pd.to_numeric(df['Discount_Percent'], errors='coerce')
```
